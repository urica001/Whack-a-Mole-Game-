# 打地鼠小游戏项目演示
项目名称：5×5打地鼠小游戏
版本Tag：v1.0
GitHub仓库链接：https://urica001.github.io/Whack-a-Mole-Game-/

## 📖项目简介
- 开局随机固定地鼠坐标，地鼠不会移动
- 玩家拥有最多5次点击机会寻找地鼠
- 点击反馈分为三类：命中、邻近（上下左右）、未命中
- 次数耗尽自动公布地鼠位置，支持一键重新开局

## ⚙️核心游戏逻辑代码
下面为游戏判断逻辑，可直接运行测试，不启动图形窗口

````python
import random

class MoleGame:
    def __init__(self):
        self.size = 5
        self.max_try = 5
        # 随机生成地鼠坐标 0‑4
        self.mole_x = random.randint(0, 4)
        self.mole_y = random.randint(0, 4)
        self.remain = self.max_try

    def check_click(self, x, y):
        """
        判断一次点击
        return: hit / near / miss
        """
        if x == self.mole_x and y == self.mole_y:
            return "hit"
        # 曼哈顿距离=1代表上下左右相邻
        elif abs(x - self.mole_x) + abs(y - self.mole_y) == 1:
            return "near"
        else:
            return "miss"
````

````python
game = MoleGame()
print(f"✅地鼠隐藏坐标: ({game.mole_x}, {game.mole_y})")

# 模拟几次点击测试
test_points = [(0,0), (game.mole_x, game.mole_y), (game.mole_x+1, game.mole_y)]
for px, py in test_points:
    res = game.check_click(px, py)
    print(f"点击({px},{py}) → {res}")
````
